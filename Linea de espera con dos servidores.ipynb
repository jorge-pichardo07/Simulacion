```python
import numpy as np
import random as r
```

En el codigo de linea de espera, tengo las variables:
- **t** es el tiempo inicial
- **NA** : Las llegadas totales al sistema
- **Nd**: Las salidas totales
Y n1, n2 son los clientes del servidor 1 y servidor 2

Los tiempos en que ocurren las cosas, para la llegada se usa el proceso de poisson, y los otros tiempos (de salida del servidor) se establecen en infinito.

Luego cree las listas para las llegadas y salidas de ambos servidores.

```python
def linea(lam1, mu1, mu2, T):
    t = 0.0
    Na = 0   
    Nd = 0   
    
    n1 = 0   
    n2 = 0   

    
    ta = -np.log(r.random()) / lam1  
    td1 = float('inf')  
    td2 = float('inf')  
    
    # Listas para guardar los tiempos 
    Llegadas_S1 = []
    Salidas_S1 = []
    Salidas_S2 = []

    while min(ta, td1, td2) < T:
        
        sig_evento = min(ta, td1, td2)
        
        #Llegada al Sistema
        if sig_evento == ta:
            t = ta
            Na += 1
            n1 += 1
            
            Llegadas_S1.append(t)
            
            # siguiente llegada
            ta = t - np.log(r.random()) / lam1
            
            # Si el Servidor 1 estaba vacío, el cliente se atiende de inmediato
            if n1 == 1:
                Y1 = -np.log(r.random()) / mu1
                td1 = t + Y1

        # salida de servidor 1 y la entrada a servidor 2
        elif sig_evento == td1:
            t = td1
            n1 -= 1  
            n2 += 1  
            
            Salidas_S1.append(t)
            
            # Actualizar el Servidor 1
            if n1 == 0:
                td1 = float('inf')
            else:
                Y1 = -np.log(r.random()) / mu1
                td1 = t + Y1
                
            # Actualizar el Servidor 2 
            if n2 == 1:
                Y2 = -np.log(r.random()) / mu2
                td2 = t + Y2

        #Salida del sistema
        elif sig_evento == td2:
            t = td2
            Nd += 1
            n2 -= 1  # Sale del sistema
    
            Salidas_S2.append(t)
            
            # Actualizar el Servidor 2
            if n2 == 0:
                td2 = float('inf')
            else:
                Y2 = -np.log(r.random()) / mu2
                td2 = t + Y2

    return Llegadas_S1, Salidas_S1, Salidas_S2
```

Aqui para probar el programa uso $\lambda_1=0.5$ y $\lambda_2=0.8$ y el tiempo total lo dejé en 20 y nos arroja los siguientes resultados:

```python
llegadas, salidas_s1, salidas_s2 = linea(0.5, 0.8, 0.7, 20)

print("Llegadas totales a S1: ", len(llegadas))
print("Salidas de S1 (Llegadas a S2): ", len(salidas_s1))
print("Salidas finales del sistema: ", len(salidas_s2))
```
