```python
import numpy as np
import random as r
from scipy import stats
```

Usaré la función vista en clase para simular las variables $x_1$ y $x_2$ con medias $\mu_1 = 1.5 $ y $\mu_2 = 1.48$ y varianzas $\sigma_1^2 = 0.16 $ y $\sigma_2^2 = 0.09$

```python
def gausiana(n , mu, sigma):
    muestra = []
    for i in range(n):
        suma = 0
        for j in range(12):
            u = r.random()
            suma += u
        x = mu + sigma*(suma-6)
        muestra.append(x)
    return muestra
```

Aquí veré el número de veces que $x_2$ supera a $x_1$ y se genera una interferencia

```python
def prob(n):
    numero=0
    p= 0
    x1 = gausiana(100000, 1.5,0.04)
    x2 = gausiana(100000,1.48,0.03)
    for i in range(n):
        if x1[i] < x2[i]:
            numero += 1
        p = numero/n
    return p
```
