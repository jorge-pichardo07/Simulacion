```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
from scipy import stats
from tabulate import tabulate
from sympy import symbols
```

```python
def funcion():
  funcion = input("Ingresa tu función de 'x' (ej: np.sin(x), x**2): ")
  return eval(f"lambda x: {funcion}", {'np': np})
```

```python
def dardos(N,funcion,ax,bx):
    exitos = 0
    pex = []
    pf = []
    a= np.linspace(ax,bx,100)
    plt.plot(a, funcion(a) ,color = 'green')
    for i in range(N):
        x = r.uniform(ax, bx)
        y = r.random()

        if y <= funcion(x):
            exitos += 1
            pex.append((x, y))
        else:
            pf.append((x, y))
    for i in range(len(pex)):
      plt.scatter(pex[i][0], pex[i][1], color = 'red')
    for i in range(len(pf)):
      plt.scatter(pf[i][0], pf[i][1], color = 'blue')
    I = (exitos/N)*bx
    P = I/bx
    print(f'El valor estimado de la integral es: {I:.2f}')
    print(f'La probabilidad es: {P:.2f}')
    return pex, pf
```

```python
veo = funcion()
a = dardos(100,veo, 0, 1)
```
