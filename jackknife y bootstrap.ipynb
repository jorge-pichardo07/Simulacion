**METODO BOOTSTRAP** \
El método bootstrap es una técnica de remuestreo que permite estimar la distribución de una estadística a partir de los datos observados, sin asumir una forma paramétrica específica.

- Remuestreo con reemplazo: Se extraen múltiples pseudomuestras de la muestra original, cada una del mismo tamaño que la muestra inicial, permitiendo que algunos datos se repitan 

- Cálculo de la estadística: Para cada pseudomuestra se calcula la estadística de interés, como la media, mediana o varianza 

- Repetición: Este proceso se repite miles de veces para generar una distribución empírica de la estadística, conocida como distribución bootstrap 

- Estimación de incertidumbre: A partir de esta distribución se pueden calcular errores estándar, sesgos y construir intervalos de confianza sin necesidad de suposiciones paramétricas

Aquí en mi función de bootstrap empiezo pidiendo los datos y el número de repeticiones, creo una lista donde se guardaran todas las estimaciones de la media de la muestra.

Dentro del ciclo donde se harán todas las repeticiones uso la función **r.choices** que sirve para escoger aleatoriamente n (en este caso 12) datos de la muestra dada y asi hacer una lista nueva de datos aleatorios, luego guardo la media de esa muestra en cada repetición del ciclo, al final solo calculo el error estándar

```python
def bootstrap(D, B):
    n = len(D)
    medias_bootstrap = []
    
    for i in range(B):
        muestra_b = r.choices(D, k=n)
        medias_bootstrap.append(np.mean(muestra_b))
        
    # El error estándar de Bootstrap es la desviación estándar de las réplicas
    error_b = np.sqrt(np.var(medias_bootstrap))
    
    return medias_bootstrap, error_b
```

Entonces haciendo 2000 repeticiones del remuestreo de bootstrap, tenemos los siguientes resultados:

```python
B = 2000
medias_b, error_b = bootstrap(datos, B)
print(f"Las media de las medias dadas por el metodo de bootsptrap es: {np.mean(medias_b)}")
print(error_b)
```

**METODO JACKKNIFE**\
es una técnica de remuestreo que se utiliza principalmente para dos cosas:

- Estimar el sesgo de un estimador estadístico.

- Calcular el error estándar (la variabilidad) de ese estimador.

El algoritmo es:

- Eliminar un dato: Crea una submuestra quitando el primer elemento. teniendo una muestra de tamaño $n-1$.
- Calcular el estadístico: Calcula el valor interesado (por ejemplo, la media, la varianza o la mediana) usando solo esos $n-1$ datos. se le llama una réplica de Jackknife.
- Repetir: Se regresa ese dato a la muestra y ahora se quita el segundo elemento. Se vuelve a calcular el estadístico.
- Iterar hasta el final: Se repite el proceso consecutivamente hasta que se haya omitido cada uno de los datos exactamente una vez.
- Al final tendremos exactamente $n$ réplicas o submuestras.

Ahora, en mi codigo de Jackknife pido solamente la lista de Datos para crear las submuestras, creo una lista con las medias que vamos a obtener de cada submuestra, aquí haré un ciclo con un rango $n$ porque tengo que quitar el elemento i en cada iteración, para eso excluyo el termino i en cada repetición y agrego lo que resta de la lista, luego guardo la media de las muestras nuevas

EL error estandar de Jackknife se calcula como 
$$
sd = \sqrt{ \frac{(n - 1)}{n} * \sum(\theta_i - \theta{\cdot})^2  }
$$

Para eso calculo el promedio de las medias guardadas y calculo el error estándar

```python

def jackknife(D):
    n = len(D)
    medias_jackknife = []
    
    for i in range(n):
        # Creamos una submuestra excluyendo el elemento en el índice 'i'
        muestra_j = D[:i] + D[i+1:]
        medias_jackknife.append(np.mean(muestra_j))
        
    # Fórmula del error estándar de Jackknife:
    # SE = sqrt( ((n - 1) / n) * sum( (theta_i - theta_punto)^2 ) )
    promedio_medias_j = np.mean(medias_jackknife)
    suma_cuadrados = sum((m - promedio_medias_j)**2 for m in medias_jackknife)
    error_estandar_jack = np.sqrt(((n - 1) / n) * suma_cuadrados)
    
    return medias_jackknife, error_estandar_jack

```

Entonces el resultado de la media y el error estándar con el método de Jackknife queda como:

```python
medias_jack, error_jack = jackknife(datos)
print(f"Las media de las medias dadas por el metodo de jacknife es: {np.mean(medias_jack)}")
print(f"y el error estandar queda como: {error_jack}")
```
