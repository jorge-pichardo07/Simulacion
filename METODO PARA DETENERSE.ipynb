**ACTIVIDAD**
Estimar la integral $$\int_0^1 e^{x^2} dx$$ por montecarlo \
Generar al menos 100 valores y deténgase cuando la desviación estandar de estimador sea menor que 0.01 y 0.001

```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
```

Uso la funcion de crudo para saber el valor de la integral

```python
def crudo(a,b,N):
  I = 0
  for i in range(N):
    x= r.random()
    I += f(x)
  return (I/N)*(b-a)
```

```python
def f(x):
  return np.exp(x**2)
```

Aqui genero el primer valor de la simulación para el valor de la integral

```python
Prueba = crudo(0, 1,1)
```

Solo visualizo el primer dato

```python
print(Prueba)
```

Aqui uso d como la desviación estandar, g como los datos que se van a generar (si se quiere modificar, pero en este caso serán 30), N será el numero de simulaciones

```python
def detener(d,g):
    Datos = []
    for i in range(g):
      D = crudo(0, 1, 1)
      Datos.append(D)

    if np.var(Datos)/np.sqrt(g)<d:
        print("Detenido, solo fue necesario usar las 30 muestras")
    else:
        k = g
        #Hago un ciclo while para obtener el mejor generador
        while (np.var(Datos)/np.sqrt(k)>d):
            D = crudo(0, 1, 1)
            Datos.append(D)
            k = k+1

    return np.mean(Datos), np.var(Datos)/np.sqrt(k), k
```

```python
Prueba = detener(0.01,30)
print(f"El número de simulaciones {Prueba[2]}")
print(f"Notemos que como {Prueba[1]:.5f} es menor que el valor 0.01")
print(f"la estimación de la integral es {Prueba[0]:.5f}, que es la media de los datos")
```
