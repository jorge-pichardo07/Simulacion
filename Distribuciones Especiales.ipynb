```python
import random as r
import numpy as np
import matplotlib.pyplot as plt
```

DISTRIBUCIÓN TRIANGULAR

```python
def triangular(n):
  l=[]
  for i in range(n):
    u = r.random()
    if 0 <= u < 0.5:
      x= np.sqrt(2*u)
      l.append(x)
    else:
      x =2-np.sqrt(2*(1-u))
      l.append(x)
  return l
m=triangular(100000)
plt.hist(m,bins=100)
plt.show()
def cdf(m):
  n=len(m)
  x=np.sort(m)
  y=np.arange(0,n)/n
  return x,y
x_math, y_math = cdf(m)
media= np.mean(m)
plt.step(x_math,y_math)
plt.axhline(y=0.5, color='red')
plt.axvline(x=media, color='green')
plt.show()
```

GUMBEL ESTANDAR

```python
def G(n):
  l=[]
  for i in range(n):
    u = r.random()
    if 0 < u < 1 :
      x=-np.log(-np.log(u))
      l.append(x)

  return l
m=G(100000)
plt.hist(m,bins=100)
plt.show()
def cdf(m):
  n=len(m)
  x=np.sort(m)
  y=np.arange(0,n)/n
  return x,y
x_math, y_math = cdf(m)
media= np.mean(m)
plt.step(x_math,y_math)
plt.axhline(y=0.5, color='red')
plt.axvline(x=media, color='green')
plt.show()
```

CAUCHY

```python
def cauchy(n):
  l=[]
  for i in range(n):
    u = r.random()
    if 0 < u < 1 :
      x=np.tan(np.pi*(u-0.5))
      l.append(x)
  return l
m= cauchy(100000)
plt.hist(m,bins=100)
plt.show()
def cdf(m):
  n=len(m)
  x=np.sort(m)
  y=np.arange(0,n)/n
  return x,y
x_math, y_math = cdf(m)
media= np.mean(m)
plt.step(x_math,y_math)
plt.axhline(y=0.5, color='red')
plt.axvline(x=media, color='green')
plt.show()
```
