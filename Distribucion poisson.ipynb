Del código visto en clase haré el código para la distibución de poisson
, esta distribución de probabilidad sirve para modelar la probabilidad de que ocurra un número determinado de eventos independientes en un intervalo fijo de tiempo o espacio, bajo una tasa constante de ocurrencia.

Usualmente estima la probabilidad de sucesos inusuales, como fallos de maquinaria o accidentes de tráfico.

Puede optimizar el flujo en bancos, servidores web o cajas de supermercado al calcular cuántos clientes llegarán por hora.

Con esta distibución tambien puede contar el número de defectos en un metro cuadrado de tela, un lote de fabricación o un cable de fibra óptica.

Importaré las libreiras que usualmente usamos

```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

En mi función poisson agregaré los parámetros n (número de simulaciones) y lambda (El lambda de parámetro), generé primero una variable aleatoria entre 0 y 1, un contador con i = 0 que será nuestras iteraciones necesarias para simular p, el ciclo while nos indica cuando detenernos en nuestra simulación para empezar de nuevo con el for

```python
def poisson(n, lam):
    muestra = []
    for x in range(n):
        u = r.random()
        i = 0
        p = np.exp(-lam)
        F = p

        while u > F:
            p = (lam * p) / (i + 1)
            F += p
            i += 1

        muestra.append(i)

    return muestra


```
