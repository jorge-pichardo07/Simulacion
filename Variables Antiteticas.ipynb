Estimación de la integral 
$$
\int_0^1 e^{x^2} dx
$$

Usando variables antitéticas

Primero importo las librerias y pongo la función que vamos a estimar

```python
import numpy as np
import random as r
```

```python
def f(x):
    return np.exp(x**2)
```

Usaré las variables anntitéticas de una distribución uniforme $u$ ~ $U(1,0)$, estoy evaluando el punto con el antitético
$$
\frac{f(U)+f(1-U)}{2}
$$
y estoy usando N//2 porque al ser el punto con su variable antitetica ya se estan haciendo 2 muestras, entonces divido N entre dos y las fos rayas divisoras son para obtener un valor entero

```python
def antiteticas(a, b, N):
    I = 0
    # Corremos hasta N//2 porque cada iteración genera 2 muestras (U y 1-U)
    for i in range(N // 2):
        u = r.random()
        # Evaluamos el punto original y su antitético (1 - u)
        I += (f(u) + f(1 - u)) / 2
        
    # Para calcular el valor de la integral, dividimos entre N//2 y multiplicamos por (b-a)
    return (I / (N // 2)) * (b - a)
```

Aquí hago una prueba para ver el valor aproximado de la integral

```python
antiteticas(0,1,50)
```

Ahora usaré la función para detener el programa en n iteraciones suficientes, así notaremos el cambio de iteraciones en la estimación con el metodo crudo y contra el de variables antiteticas

```python
def detener(d, g):
    Datos = []
    for i in range(g):
        # Usamos 2 muestras (1 par antitético) para mantener la estructura equivalente
        D = antiteticas(0, 1, 2) 
        Datos.append(D)

    if np.var(Datos) / np.sqrt(g) < d:
        k=g
        print(f"Detenido, solo fue necesario usar las {k} muestras iniciales")    
    else:
        k = g
        while (np.var(Datos) / np.sqrt(k) > d):
            D = antiteticas(0, 1, 2)
            Datos.append(D)
            k = k + 1

    return np.mean(Datos), np.var(Datos) / np.sqrt(k), k
```

```python
# Parámetros de prueba
tol = 0.005
muestras = 30

# Ejecutar estimación
resultado, error, total_k = detener(tol, muestras)

print(f"Valor estimado de la integral: {resultado:.6f}")
print(f"Varianza/error final: {error:.6f}")
print(f"Total de iteraciones del ciclo (k): {total_k}")
```

Notemos que para el metódo de variables antitéticas solo usa las 30 muestras iniciales

Comparando con el metodo de MC crudo

```python
def crudo(a,b,N):
  I = 0
  for i in range(N):
    x= r.random()
    I += f(x)
  return (I/N)*(b-a)
```

```python
def detener1(d,g):
    Datos = []
    for i in range(g):
      D = crudo(0, 1, 1)
      Datos.append(D)

    if np.var(Datos)/np.sqrt(g)<d:
        k=g
        print(f"Detenido, solo fue necesario usar las {g} muestras")
    else:
        k = g
        #Hago un ciclo while para obtener el mejor generador
        while (np.var(Datos)/np.sqrt(k)>d):
            D = crudo(0, 1, 1)
            Datos.append(D)
            k = k+1

    return np.mean(Datos), np.var(Datos)/np.sqrt(k), k
```

```python
# Parámetros de prueba
tol = 0.005
muestras = 30

# Ejecutar estimación
resultado, error, total_k = detener1(tol, muestras)

print(f"Valor estimado de la integral: {resultado:.6f}")
print(f"Varianza/error final: {error:.6f}")
print(f"Total de iteraciones del ciclo (k): {total_k}")
```

Ahora vemos que para el método de montecarlo crudo fueron necesarias 2000 iteraciones, comprobamos el uso de las variables antitéticas.
