```python
import numpy as np
import random as r
```

Simulación del tiempo de llegada generamos distribuciones exponenciales

```python
def TL(t_actual, lam):
    V = r.random()
    return t_actual - (np.log(V) / lam)
```

```python
def TS(mu):
    V = r.random()
    return -(np.log(V) / mu)
```

En esta parte del código, estoy agregando el tiempo en cero, el numero de arrivos en cero, el numero de salidas y los clientes que van apareciendo

Luego, empiezo el caso uno con un while True, caso 1 donde el próximo evento es una llegada y ocurre antes del tiempo límite T, el caso 2 el próximo evento es una salida y ocurre antes o en el tiempo límite T, el caso 3 ya pasó el tiempo límite T, pero todavía quedan clientes en el sistema y en el caso 4 ya pasó el tiempo límite T y el sistema se ha vaciado por completo

```python
def Linea_Espera(lam, mu, T):

    t = 0.0
    N_A = 0
    N_S = 0
    n = 0

    
    A = {}  
    D = {}  

    # Generar la primera llegada T_0 y configurar lista de eventos (LE)
    t_A = TL(t, lam)
    t_D = float("inf")  


    while True:
        if t_A <= t_D and t_A <= T:
            t = t_A
            N_A += 1
            n += 1  #agrego un cliente

            #uso la funcion para la hora de la siguiente llegada
            t_A = TL(t, lam)

            # si no habia nadie el cliente pasa directo a servicio
            if n == 1:
                Y = TS(mu)
                t_D = t + Y

            # GUardo el dato de salida
            A[N_A] = t

        elif t_D < t_A and t_D <= T:
            t = t_D
            n -= 1  #cliente salió
            N_S += 1  # contador de gente q sale
            D[N_S] = t

            # cliente por atender
            if n == 0:
                t_D = float("inf")
            else:
                Y = TS(mu)
                t_D = t + Y

        elif min(t_A, t_D) > T and n > 0:
            t = t_D
            n -= 1
            N_S += 1

            D[N_S] = t

            if n > 0:
                Y = TS(mu)
                t_D = t + Y

        elif min(t_A, t_D) > T and n == 0:
            # Calcular el tiempo extra que el servidor se quedó después de T
            T_p = max(t - T, 0)
            break

    tiempos_en_sistema = [D[i] - A[i] for i in range(1, N_S + 1)]
    tiempo_promedio_sistema = (
        sum(tiempos_en_sistema) / N_S if N_S > 0 else 0.0
    )

    return {
        "N_A": N_A,
        "N_D": N_S,
        "Tiempo_Promedio_Sistema": tiempo_promedio_sistema,
        "T_p": T_p,
    }
```

veamos la solución de la simulación

```python
lam = 4
mu = 6
corte = 100

resultado = Linea_Espera(lam, mu, corte)

print(f"Clientes que llegaron (N_A): {resultado['N_A']}")
print(f"Clientes que salieron (N_D): {resultado['N_D']}")
print(
    f"El tiempo promedio de un cliente en el sistema: {resultado['Tiempo_Promedio_Sistema']:.4f} unidades de tiempo"
)
print(
    f"Tiempo extra del servidor después de T (T_p): {resultado['T_p']:.4f} unidades de tiempo"
)
```
