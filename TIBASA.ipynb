La empresa TIBASA (Fabricante de tinas de baño) tiene asignado un camión especial para el transporte de tinas terminadas. Dicho camión transporta diariamente 5 tinas. El peso de cada tina sigue la distribución triangular de la imagen anexa. 

**PASO 1**
***
Definir las variables\
**Variables**: Tinas de diferentes pesos

Importamos las librerías

```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

Generamos las suma de los pesos de las tinas, porque tienen una distribución triangular

```python
def tinas(n):
    TINAS = []
    x = 0
    for i in range(n):
        for j in range(5):
            x += r.triangular(190,230,210)
        
        TINAS.append(x)
        x= 0


    return TINAS
```

Verificamos que la función, nos dá el peso total de las 5 tinas, para cada repetición

```python
PRUEBA= tinas(2)
print(f'la primer observación nos da un peso total de {PRUEBA[0]:.2f}',' \n sabemos que es correcto, porque el peso maxímo es 1150')
```

Ahora, ¿cuál es la probabilidad de que el peso de las tinas exceda la capacidad del camión?, para eso hare una simulación donde si el peso de la suma de las 5 tinas excede la capacidad del camión sume los casos posibles en los casos totales

```python
def prob(n):
    T = tinas(n) 
    favorables = 0
    for i in range(n):
        if T[i] > 1000:
            favorables += 1
    probabilidad = favorables/n

    return probabilidad


```

Vamos a realizar 200 observaciones, para tener el número de casos favorables (**EXCEDE LA CARGA**), entre n que son nuestros casos posibles

```python
Probabilidad = prob(200)
```

```python
print('Con la simulación obtuvimos que la probabilidad de\n',f'que el peso de las tinas exceda la capacidad del camión es {Probabilidad:.3f}')
```
